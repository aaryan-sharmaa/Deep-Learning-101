## Creating a Tensor

```python
# using empty
x=torch.empty(2,3)
print(x)
```

```python
# check type
print(type(x))
```

```python
# using zeros
x=torch.zeros(2,3)
print(x)
```

```python
# using ones
x=torch.ones(2,3)
print(x)
```

```python
# using rand
x=torch.rand(2,3)
print(x)
```

```python
# use of seed
torch.manual_seed(100)
x=torch.rand(2,3)
print(x)
```

```python
# using tensor
x=torch.tensor([[1,2,3],[4,5,6]])
print(x)
```

### 2. Element wise operations

```python
a = torch.rand(2,3)
b = torch.rand(2,3)

print(a)
print(b)
```

```python
# add
torch.add(a,b)

# subtract
torch.sub(a,b)

# multiply
torch.mul(a,b)

# divide
torch.div(a,b)

# mod
torch.remainder(a,b)

# power
torch.pow(a,b)
```

### 3. Reduction **Operation**

```python
e = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
e
```

```python
# sum
torch.sum(e)

#sum along columns
torch.sum(e, dim=0)

#sum along rows
torch.sum(e, dim=1)
```

```python
# mean
torch.mean(e)

#mean along columns
torch.mean(e, dim=0)

#mean along rows
torch.mean(e, dim=1)
```

```python
#median
torch.median(e)

#median along columns
torch.median(e, dim=0)

#median along rows
torch.median(e, dim=1)
```

```python
 # max and min
torch.max(e)
torch.min(e)
```

```python
# product
torch.prod(e)
```

```python
# standard deviation
torch.std(e)
```

```python
# variance
torch.var(e)
```

```python
# argmax
torch.argmax(e)
```

```python
# argmin
torch.argmin(e)
```

### 4. Matrix Operations

```python
f = torch.randint(size=(2,3), low=0, high=10)
g = torch.randint(size=(3,2), low=0, high=10)

print(f)
print(g)
```

```python
# matrix multiplication
torch.mm(f,g)
```

```python
vector1 = torch.tensor([1, 2])
vector2 = torch.tensor([3, 4])

# dot product
torch.dot(vector1, vector2)
```

```python
# transpose
torch.t(f)
```

```python
h = torch.randint(size=(3,3), low=0, high=10, dtype=torch.float32)
h
```

```python
# determinant
torch.det(h)
```

```python
# inverse
torch.inverse(h)
```

### 5. Comparison Operations

```python
i = torch.randint(size=(2,3), low=0, high=10)
j = torch.randint(size=(2,3), low=0, high=10)

print(i)
print(j)
```

```python
# greater than
torch.gt(i,j)
```

```python
# lesser than
torch.lt(i,j)
```

```python
# equal to
torch.eq(i,j)
```

```python
# not equal to
torch.ne(i,j)
```

```python
# less than equal to
torch.le(i,j)
```
